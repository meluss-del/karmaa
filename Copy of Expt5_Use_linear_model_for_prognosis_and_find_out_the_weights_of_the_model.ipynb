```python
# Import the module 'LinearRegression' from sklearn
from matplotlib import pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Create an object of type LinearRegression
model = LinearRegression()
model
```

```python
import numpy as np
import pandas as pd

def load_data(n=100):
    np.random.seed(0)

    age = np.random.randint(20, 70, n)
    cholesterol = np.random.randint(150, 300, n)

    X = pd.DataFrame({
        'Age': age,
        'Cholesterol': cholesterol
    })

    y = 0.5 * age + 0.3 * cholesterol + np.random.randn(n) * 10

    return X, y
```

```python
# Generate features and labels using the imported function
X, y = load_data(100)
```

```python
# View the features
X.head()
```

```python
# Plot a histogram of the Age feature
plt.xlabel('Age')
X['Age'].hist();
```

```python
# Plot a histogram of the cholesterol feature
plt.xlabel('Cholesterol')
X['Cholesterol'].hist();
```

```python
#Fit the linear regression model
model.fit(X, y)
model
```

```python
# View the coefficients (weights) of the model
model.coef_
```
