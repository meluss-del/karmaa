```python
import numpy as np
import pandas as pd

def load_data(n=100):
    np.random.seed(0)

    age = np.random.randint(20, 70, n)

    # Generate blood pressure data
    diastolic_bp = 55 + 0.25 * age + np.random.randn(n) * 4
    systolic_bp = 95 + 0.75 * age + np.random.randn(n) * 6

    X = pd.DataFrame({
        'Age': age,
        'Diastolic_BP': diastolic_bp,
        'Systolic_BP': systolic_bp
    })

    # Target variable
    y = systolic_bp

    return X, y
```

```python
# Generate features and labels using the imported function
X, y = load_data(100)
```
